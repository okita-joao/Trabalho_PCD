# **Código de criação de grafos experimentais**

 Tal código foi utilizado para gerar grafos não-direcionados que devem ser utilizados como base para a experimentação da implementação da Busca em Largura de modo serial e concorrente.

###**Instruções para gerar um grafo de experimentação**

1. Defina o número de vértices do grafo não-direcionado alterando o valor da variável *n*.

2. Indique se o grafo a ser gerado deve ser completo ou não pela variável de sinalização *completo*.

3. Defina o caminho do arquivo que conterá a matriz de adjacência do grafo pela variável *file_name*.



```python
import numpy as np
import csv

n = 100
completo = False
file_name = f"grafo_{n}x{n}.csv"

matriz_adj = np.full((n, n), 1)

if(not completo):
  # Criando uma matriz de adjacências n x n simétrica (Grafo Não-Direcionado)
  for i in range(0, n):
      arestas = np.random.choice([0, 1], size = n - i)

      for j in range(i, n):
        x = arestas[j - i]
        matriz_adj[i, j] = x
        matriz_adj[j, i] = x

# Prevenção de laços no grafo
for i in range(n):
  matriz_adj[i, i] = 0

# Convertendo os valores da matriz para int de 8bits
matriz_adj = matriz_adj.astype(np.int8)

# Escrevendo o grafo gerado num arquivo .csv
with open(f"grafo_{n}x{n}.csv", 'w', newline='') as csvfile:
  grafo_writer = csv.writer(csvfile, delimiter=',')
  grafo_writer.writerows(matriz_adj)

```
